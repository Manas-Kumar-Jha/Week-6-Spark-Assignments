## Q3: Read CSV with header and infer schema

```python
df = spark.read.option("header",True).option("inferSchema",True).csv("data/source.csv")
df.show(5)
df.printSchema()
```

## Q5: Select product_id and price where category is 'Electronics'

```python
result = df.filter(col("category")=="Electronics").select("product_id","price")
result.show()
```
